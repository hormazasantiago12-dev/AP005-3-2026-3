```markdown
Este bloque de código importa las librerías `random` y `matplotlib.pyplot` para generar números aleatorios y visualizarlos en un gráfico de líneas, respectivamente. Luego, crea dos listas: `numbers_a` con números del 1 al 12, y `numbers_b` con 12 números enteros aleatorios entre 1 y 1000. ````python
```
````python
import random
from matplotlib import pyplot as plt

# Add your code below:
numbers_a = range(1, 13)
numbers_b = [random.randint(1, 1000) for i in range(12)]
plt.plot(numbers_a, numbers_b)
plt.show()
````
