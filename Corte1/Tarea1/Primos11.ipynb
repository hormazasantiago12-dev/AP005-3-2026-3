Este código Python calcula y muestra los números primos entre 1 y 30. También mide el tiempo de ejecución del proceso para encontrar estos números primos.

```python
import time
inicio = time.time()

for i in range(1,31):
    conta = 0
    for n in range(1, i+1):
        residue = i%n
        if residue == 0:
            conta = conta + 1
    if conta == 2:
        print(f'{i} es un primo')
        print("\n")

fin = time.time()
print("t = ", (fin - inicio)*1000)
```
