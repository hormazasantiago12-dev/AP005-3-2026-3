# Regresión Lineal Simple (Sin Sesgo / Bias)

Este notebook implementa un modelo de regresión lineal sin término independiente (sesgo), es decir, utilizando únicamente la ecuación:
$$\hat{y} = w \cdot x$$

Está escrito enteramente en **Python puro** para ilustrar de forma matemática y visual (época por época) cómo funciona el descenso de gradiente mediante derivadas parciales.

## 1. Importación y Configuración Inicial
Definimos los datos de entrenamiento (`inputs` y `targets`), el peso inicial del modelo ($w$), el número de épocas, la tasa de aprendizaje y el tiempo de pausa.

```python
import time

# Datos de entrenamiento
inputs = [1, 2, 3, 4]
targets = [2, 4, 6, 8]

# Peso inicial
w = 0.1

# Numero de epocas
epochs = 10

# Tasa de aprendizaje
learning_rate = 0.1

# Tiempo de pausa entre epocas
pause_time = 1
```

## 2. Función de Predicción y Bucle de Entrenamiento
En cada iteración o época:
1. Se realizan las predicciones con el modelo actual.
2. Se calcula el error cuadrático y el costo promedio.
3. Se calculan las derivadas del error respecto a la salida ($dE/dy$) y respecto al peso ($dE/dw$).
4. Se actualiza el peso empleando el gradiente promedio y la tasa de aprendizaje.

```python
# Funcion de prediccion
# Modelo: y = w*x
def predict(i):
    return w * i

# Entrenamiento
for epoch in range(epochs):

    # Predicciones para todos los datos
    predictions = [predict(i) for i in inputs]

    # Error cuadratico para cada dato
    errors = [
        (p - t)**2
        for p, t in zip(predictions, targets)
    ]

    # Costo promedio
    cost = sum(errors) / len(targets)

    # Derivada del error con respecto a la salida
    errors_d = [
        2 * (p - t)
        for p, t in zip(predictions, targets)
    ]

    # Derivada del error con respecto al peso
    weight_d = [
        e * i
        for e, i in zip(errors_d, inputs)
    ]

    # Mostrar informacion de la epoca
    print("\n" + "=" * 65)
    print(f"EPOCA {epoch + 1}")
    print("=" * 65)

    for i, t, p, e, ed, wd in zip(
        inputs,
        targets,
        predictions,
        errors,
        errors_d,
        weight_d
    ):
        print(
            f"Entrada: {i:>2} | "
            f"Target: {t:>2} | "
            f"Pred: {p:>8.4f} | "
            f"Error^2: {e:>8.4f} | "
            f"dE/dy: {ed:>8.4f} | "
            f"dE/dw: {wd:>8.4f}"
        )

    print("-" * 65)
    print(f"Peso antes de actualizar: {w:.10f}")
    print(f"Costo promedio: {cost:.6f}")

    # Gradiente promedio respecto al peso
    gradient = sum(weight_d) / len(weight_d)

    print(f"Gradiente promedio: {gradient:.6f}")

    # Actualizacion del peso
    w -= learning_rate * gradient

    print(f"Peso despues de actualizar: {w:.10f}")

    # Pausa didactica entre epocas
    time.sleep(pause_time)
```

## 3. Prueba Final del Modelo
Evaluamos el rendimiento del modelo ya entrenado con datos nuevos (`test_inputs = [5, 6]`) para verificar si predice correctamente los valores esperados.

```python
# Datos de prueba
test_inputs = [5, 6]
test_targets = [10, 12]

print("\n" + "=" * 65)
print("PRUEBA FINAL")
print("=" * 65)

test_predictions = [
    predict(i)
    for i in test_inputs
]

for i, t, p in zip(
    test_inputs,
    test_targets,
    test_predictions
):
    print(
        f"Entrada: {i:>2} | "
        f"Target: {t:>2} | "
        f"Prediccion: {p:>8.4f}"
    )
```
